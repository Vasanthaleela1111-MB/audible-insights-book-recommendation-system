```python
import re

df['Listening_Time_Minutes'] = df['Listening Time'].apply(
    lambda x: sum(
        int(num) * (60 if unit.startswith('hour') else 1)
        for num, unit in re.findall(r'(\d+)\s*(hour|hours|minute|minutes)', str(x))
    )
)
df.head()
```

```python
df['Rank']=df['Ranks and Genre'].str.extract(r'#(\d+)')
df['Rank']=pd.to_numeric(df['Rank'],errors='coerce')

def extract_genres(text):
    genres=re.findall(r'#\d+\s+in\s([^,#]+)', str(text))

    if len(genres)>1:
        genres=genres[1:]
    return ", ".join(genres)

df['Genre']=df['Ranks and Genre'].apply(extract_genres)
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemma=WordNetLemmatizer()

def preprocessing(text):
    if pd.isna(text):
        return ""
    text=str(text).lower()
    text=re.sub(r"[^a-zA-Z\s]", " ", text)
    words=word_tokenize(text)
    words=[lemma.lemmatize(word) for word in words]
    return " ".join(words)
```

```python
df['Book Name']=df['Book Name'].apply(preprocessing)
df['Description']=df['Description'].apply(preprocessing)
df['Genre']=df['Genre'].apply(preprocessing)
```

```python
df['Author']=df['Author'].str.lower()
```

```python
df = df.drop_duplicates(
    subset=["Book Name", "Author"]
).reset_index(drop=True)
```

```python
df['content'] = (
    df['Book Name'].astype(str) + ' ' +
    df['Author'].astype(str) + ' ' +
    df['Genre'].astype(str) + ' ' +
    df['Description'].astype(str)
)
```

```python
df['content'].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf=TfidfVectorizer(max_features=5000, stop_words='english')
tfidf_matrix=tfidf.fit_transform(df['content'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

content_similarity = cosine_similarity(tfidf_matrix)
content_similarity.shape
```

```python
import pickle

pickle.dump(content_similarity,open("books.pkl", "wb"))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Rating_norm',
    'Reviews_norm',
    'Rank_norm']] = scaler.fit_transform(
        df[['Rating',
            'Number of Reviews',
            'Rank']]
)
```

```python
df['Rank_norm'] = 1 - df['Rank_norm']
```

```python
df['Popularity_Score'] = (
    0.5 * df['Rating_norm']
    +
    0.3 * df['Reviews_norm']
    +
    0.2 * df['Rank_norm']
)
```

```python
def hybrid_recommend(book_name, top_n=5):

    matches = df[df['Book Name'].str.contains(book_name, case=False, na=False)]

    if matches.empty:
        print("Book not found!")
        return

    idx = matches.index[0]

    content_scores = content_similarity[idx]

    scores_df = pd.DataFrame({
        'Index': df.index,
        'Content_Score': content_scores,
        'Popularity_Score': df['Popularity_Score']
    })

    scores_df['Hybrid_Score'] = (
        0.7 * scores_df['Content_Score']
        + 0.3 * scores_df['Popularity_Score']
    )

    recommendations = scores_df.sort_values(
        by='Hybrid_Score',
        ascending=False
    ).iloc[1:top_n+1]

    return df.iloc[recommendations['Index']][['Book Name', 'Author']]
```

```python
def recommend(book_name):

    index = df[
        df['Book Name'].str.contains(
            book_name,
            case=False,
            na=False
        )
    ].index[0]

    distances = content_similarity[index]

    books = sorted(
        list(enumerate(distances)),
        key=lambda x:x[1],
        reverse=True
    )[1:6]

    return [
        df.iloc[i[0]]['Book Name']
        for i in books
    ]
```
